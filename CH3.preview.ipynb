#### 3-2 인덱스에 의미부여하여 풀어보기   
보통 리스트를 사용할 때는 인덱스로 데이터에 접근한다.    
그런데 상황에 따라 인덱스에 해싱(hashing)개념으르 적용하여 단순한 위치가 아니라 특정한 의미를 지닌 값으로 활용할 수도 있다.    
그 중에서도 인덱스를 순서가 아니라 해당 숫자 값 자체에 의미를 부여하는 상황을 가장 자주 사용한다.    
- A[1]의 의미    
1)몇 번째 데이터인지 순서를 의미하는 경우 : 첫 번째 데이터를 저장.    
2)숫자값으로 의미를 부여한 경우 : 1이라는 값이 몇 개 있는지를 저장   

```python
#1000보다 작은 자연수 10,000,000개를 1초 안에 정렬해야 하는 문제.
#인덱스에 의미를 부여한 사례 : 계수 정렬 

import sys

N = int(sys.stdin.readlind())
count = [0] * 1001
numbers = list(map(int, sys.stdin.readline().split()))

for number in numbers: 
    count[number] += 1 #인덱스에 숫자값으로 의미를 부여하여 데이터를 저장 

for i in range(1001):
    if count[i] != 0:
        for _ in range(count[i]):
            sys.stdout.write(str(i) + ' ')

##이 코드에서 count리스트의 인덱스는 숫자 자체이고, 값은 해당 숫자의 등장 횟수를 의미한다. 
##이렇게 하면 비교 연산을 따로 하지 않아도 10,000,000개의 숫자를 1초 안에 빠르게 정렬할 수 있다. 
##인덱스를 단순히 '몇 번째 순서'로만 생각하지 말고, 문제 상황에 따라 다양한 의미를 변환해 보는 연습이 중요하다. 
#12장 동적 계획법 참고. 
```

#### 3-3 나머지 연산의 중요성 알아보기      
분배법칙 : 나머지 연산은 나눗셈을 제외하고, 덧셈/뺄셈/곱셈의 분배법칙이 성립한다.     
(20+6) % 3 = (20 % 3) + (6 % 3) = 2    
(20-6) % 3 = (20 % 3) - (6 % 3) = 2   
(20*6) % 3 = (20 % 3) * (6 % 3) = 0   
(20/6) % 3 != (20 % 3) / (6 % 3)     
파이썬에서는 int형이 자동으로 확장되므로 정수 오버플로는 발생하지 않지만, 숫자가 커지면 계산속도가 급격히 느려져 시간초과에 걸릴 수 있다.    
즉, 마지막에만 %연산을 적용해도 계산은 되지만, 중간과정마자 나머지 연산을 적용하는 습관을 가지면 속도가 빨라진다. 

```python
import time 

answer = 1
start = time.time()

for i in range(1, 100001):
    answer *= i

result = answer % 1000000007
end = time.time()
print("결과:", result)
print("수행시간: {:.6f}초".format(end - start))
```

```python
import time 

answer = 1
start = time.time()

for i in range(1, 100001):
    answer = (answer * i) % 1000000007

end = time.time()
print("결과:", answer)
print("수행 시간: {: .6f}초".format(end-start))
```

#### 3-4 정렬 기초 다지기 
코딩 테스트는 다양한 데이터를 효과적으로 다루는 것에서 시작.    
데이터를 효과적으로 다루는 방법 중에서 정렬은 거의 모든 알고리즘 문제의 핵심 요소!     
예를 들어 이진 탐색은 정렬된 데이터에만 적용 가능.     
그리디 알고리즘, 투 포인터, 우선순위 큐 문제의 전처리에서 반드시 등장    

**오름차순 정렬**      
작은 값부터 시작해서 점차 큰 값으로 나열하는 과정     
list.sort() : 리스트 객체 자체를 정렬하며, 반환값은 None     
sorted(list) : 원본을 유지하고 정렬된 복사본을 반환     

**내림차순 정렬**     
큰 값부터 시작해서 작은 값으로 나열하는 과정     
reverse = True 옵션을 사용하면 됨    

```python
#오름차순 정렬 
#원본 리스트 선언
A = [5, 3, 2, 4, 1]

#방법1 : list.sort() - 리스트 자체를 정렬(in-place)
A.sort()
print("sort() 결과:", A)

#원본 리스트 다시 초기화
A = [5, 3, 2, 4, 1]

#방법2 : sorted(list) - 정렬된 복사본을 반환하고 원본은 유지
B = sorted(A)
print("원본 리스트", A)
print("sorted() 결과:", B)
```

```python
#내림차순 정렬 
#원본 리스트 선언
A = [5, 3, 2, 4, 1]

#리스트 자체를 내림차순 정렬
A.sort(reverse=True)
print(A)

#원본 리스트 초기화 
A = [5, 3, 2, 4, 1]

#정렬된 복사본 반환하고 원본유지 
B = sorted(A, reverse=True)
print(A)
print(B)
```

```python
#정렬 함수를 직접 제어할 수 없는 경우, 부호를 일시적으로 반전시키는 방법 

A = [5, 3, 2, 4, 1]

#부호를 반전시키고 오름차순으로 정렬한 후, 다시 부호 되돌리기 
A = [-x for x in A]  #[-5, -3, -2, -4, -1]
A.sort()             #[-5, -4, -3, -2, -1]
A = [-x for x in A]  #[5, 4, ,3 ,2 ,1]

print("부호 반전 방식 내림차순:", A)
```

#### 3-5 다중 조건 정렬 익히기   

튜플 기반 정렬은 코드 길이가 짧고 간단 -> 단순 입력 처리에 최적이고 빠르다.      
딕셔너리 기반 정렬은 구조화되어 가독성이 좋다 -> 필드명이 있는 복잡한 데이터처리에 적합하다.  

```python
#튜플 기반 정렬 : 데이터를 (기준1, 기준2, ...) 형태의 튜플로 구성한 뒤, sort() 또는 sorted() 함수의 key 인자에 lambda를 사용해 정렬 기준을 지정하는 것    
#영어, 수학 순서로 튜플 데이터 저장 
scores = [
    (80, 100),
    (100, 50),
    (70, 100),
    (80, 90)
]

#영어 내림차순, 영어가 같으면 수학 내림차순  
#scores.sort(key=lambda x: (-x[0], -x[1]))
scores.sort(key=lambda x: (x[0], x[1]), reverse=True)

print(scores)

for s in scores:
    print(f"english={s[0]}, math={s[1]}")
```

```python
#딕셔너리 기반 정렬 
#영어, 수학 순서로 튜플 데이터 저장 
scores = [
    {'english': 80, 'math': 100},
    {'english': 100, 'math': 50},
    {'english': 70, 'math': 100},
    {'english': 80, 'math': 90}
]

#수학 내림차순 정렬, 수학이 같으면 영어 내림차순 
scores.sort(key=lambda x: (-x['math'], -x['english']))
#scores.sort(key=lambda x: (x['math'], x['english']), reverse=True)

for s in scores:
    print(s)
```

#### 3-6 이차원 리스트 제대로 다루기 
이차원리스트는 그래프의 구조를 표현하는 데 많이 사용된다.       

**이차원 리스트를 이용한 그래프 구현 3단계**
1) 이차원 리스트 선언과 초기화 : list in list 
2) 그래프 데이터 저장하기 
3) 그래프 데이터 가져오기 

```python
#1단계: 이차원 리스트 선언과 초기화 
#list in list 방식으로 이차원 리스트 구성 시, 노드 번호는 1번부터 시작하므로 인덱스 0 사용하지 않고 N+1 크기로 리스트 초기화

#정점이 3개 있다고 가정(1~3번 사용)
N = 3
graph = [[] for _ in range(N+1)]

print(graph)
```

```python
#2단계: 그래프 데이터 저장하기 : 연결 정보는 (도착 노드, 가중치) 형태의 튜플로 저장

#정점 수와 간선 수 입력
N, E = 3, 4

#간선 정보
s = [1, 2, 1, 3]
e = [2, 1, 3, 2]
w = [4, 10, 7, 6]

for i in range(E):
    graph[s[i]].append((e[i], w[i]))

print(graph)
```

```python
#2단계: 그래프 데이터 저장하기 : 연결 정보는 (도착 노드, 가중치) 형태의 튜플로 저장

#정점 수와 간선 수 입력
N, E = map(int, input().split())

#간선 정보
for _ in range(E):
    s, e, w = map(int, input().split())
    graph[s].append((e, w))

print(graph)
```

```python
#3단계 : 그래프 데이터 가져오기

for nextNode, weight in graph[2]:
    print(f"next Node {nextNode}, weight = {weight}") 
```
