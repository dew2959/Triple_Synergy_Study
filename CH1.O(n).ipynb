```python
#O(N)
import random 
findNumber = random.randrange(1,101)

for i in range(1, 101):
    if i == findNumber:
        print(i)
        break
```

```python
#for 문 2개를 병렬로 넣었으니까 2n 정도일 줄 알았는데 진짜 오래걸림. 뭔가 틀린듯. 
Total_num =0
num = []

input(Total_num)
for i in range(Total_num):
    num.append[input(num)]

num.sort()
for i in num:
    print(i)
    print("\n")
```
